```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 2行×3列、6個のサブプロットを作成する
fig, axes = plt.subplots(2, 3, tight_layout=True)

# 折れ線グラフ
y0 = [1, 2, -5, 2]
axes[0,0].plot(y0)

# sin関数
x1 = np.linspace(0, 10, 100)
y1 = 2 + 2 * np.sin(2 * x1)
axes[0,1].set(xlim=(0, 10), xticks=np.arange(0, 10),
              ylim=(-1, 5), yticks=np.arange(-1, 6))
axes[0,1].plot(x1, y1)

# ヒストグラム（平均15、標準偏差5の正規分布から2000個の乱数を生成する）
x2 = np.random.normal(15, 5, 2000)
axes[0,2].hist(x2, bins=30)
axes[0,2].set(xlim=(-10, 40), xticks=np.arange(-10, 41, 10),
              ylim=(0, 200), yticks=np.arange(0, 201, 50))

# 散布図
np.random.seed(3)
x3 = 4 + np.random.normal(0, 2, 60)
y3 = 4 + np.random.normal(0, 2, len(x3))
axes[1,0].scatter(x3, y3)

# グラフを表示する
plt.show()
```
